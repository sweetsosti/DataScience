```python
import pandas as pd
import numpy as np
import seaborn as sns
from tensorflow import keras
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('mushrooms.csv')
dataset.head()
```

```python
sns.set(style="whitegrid")

# Count plot to visualize the distinction
plt.figure(figsize=(8, 6))
sns.countplot(data=dataset, x='class', palette='Set2')

# Add labels and title
plt.title('Distribution of Edible and Poisonous Mushrooms', fontsize=16)
plt.xlabel('Class (Edible or Poisonous)', fontsize=12)
plt.ylabel('Count', fontsize=12)

# Annotate counts
for p in plt.gca().patches:
    plt.text(p.get_x() + p.get_width() / 2., p.get_height() + 50,
             f'{int(p.get_height())}', ha='center', fontsize=10)

plt.tight_layout()
plt.show()
```
